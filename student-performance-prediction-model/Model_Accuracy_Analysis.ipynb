## 1. Setup and Imports

Import all required libraries for model evaluation and visualization.

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import os
import sys
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import (
    mean_absolute_error, 
    mean_squared_error, 
    r2_score,
    mean_absolute_percentage_error
)

# Add parent directory to path for config imports
sys.path.append(os.path.dirname(os.path.dirname(os.path.abspath('__file__'))))

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

print("All libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
print(f"Seaborn version: {sns.__version__}")
print("\n📊 Model: RandomForestRegressor (improved from LinearRegression)")
print("🔧 Encoder: OneHotEncoder (improved from LabelEncoder)")
```

## 2. Load Model and Data

Load the trained model, scaler, and test dataset for evaluation.

```python
# Configuration paths
MODEL_DIR = 'models'
DATA_DIR = 'data'
CONFIG_DIR = 'config'

# File paths (UPDATED for new model structure)
MODEL_PATH = os.path.join(MODEL_DIR, 'performance_predictor.pkl')
SCALER_PATH = os.path.join(MODEL_DIR, 'scaler.pkl')
ENCODER_PATH = os.path.join(MODEL_DIR, 'subject_encoder.pkl')  # OneHotEncoder (was label_encoder.pkl)
FEATURE_ORDER_PATH = os.path.join(MODEL_DIR, 'feature_order.pkl')  # NEW: for consistent feature ordering
DATA_PATH = os.path.join(DATA_DIR, 'cleaned_data.csv')

print("Checking file paths...")
print(f"Model path: {MODEL_PATH}")
print(f"Scaler path: {SCALER_PATH}")
print(f"Subject Encoder path: {ENCODER_PATH}")
print(f"Feature Order path: {FEATURE_ORDER_PATH}")
print(f"Data path: {DATA_PATH}")

# Check if files exist
files_exist = {
    'Model': os.path.exists(MODEL_PATH),
    'Scaler': os.path.exists(SCALER_PATH),
    'Subject Encoder': os.path.exists(ENCODER_PATH),
    'Feature Order': os.path.exists(FEATURE_ORDER_PATH),
    'Data': os.path.exists(DATA_PATH)
}

for file_name, exists in files_exist.items():
    status = "✅ Found" if exists else "❌ Missing"
    print(f"{file_name}: {status}")

if not all(files_exist.values()):
    print("\n⚠️ Some files are missing. Please ensure you have:")
    print("1. Trained model (run: python src/model_trainer.py)")
    print("2. Cleaned dataset (run: python src/data_preprocessing.py)")
    raise FileNotFoundError("Required files not found")
```

```python
# Load the trained model and preprocessing objects
print("Loading trained model and preprocessing objects...")

try:
    # Load RandomForest model (UPGRADED from LinearRegression)
    model = joblib.load(MODEL_PATH)
    print(f"✅ Model loaded: {type(model).__name__}")
    
    # Load scaler
    scaler = joblib.load(SCALER_PATH)
    print(f"✅ Scaler loaded: {type(scaler).__name__}")
    
    # Load OneHotEncoder (UPGRADED from LabelEncoder)
    subject_encoder = joblib.load(ENCODER_PATH)
    print(f"✅ Subject Encoder loaded: {type(subject_encoder).__name__}")
    
    # Load feature order for consistent predictions
    feature_order = joblib.load(FEATURE_ORDER_PATH)
    print(f"✅ Feature order loaded: {len(feature_order)} features")
    
    # Display model info
    print(f"\n📊 Model Configuration:")
    print(f"  - Type: {type(model).__name__}")
    if hasattr(model, 'n_estimators'):
        print(f"  - Number of trees: {model.n_estimators}")
    if hasattr(model, 'max_depth'):
        print(f"  - Max depth: {model.max_depth}")
    
    print(f"\n📝 Subject Encoding (One-Hot):")
    print(f"  Categories: {list(subject_encoder.categories_[0])}")
    
except Exception as e:
    print(f"❌ Error loading model files: {e}")
    raise
```

```python
# Load and prepare the dataset
print("Loading and preparing dataset...")

try:
    # Load data
    df = pd.read_csv(DATA_PATH)
    print(f"✅ Dataset loaded: {len(df)} records, {len(df.columns)} columns")
    
    # Display basic info
    print("\n📋 Dataset Overview:")
    print(f"Columns: {list(df.columns)}")
    
    # Check for engineered features
    engineered_features = ['attendance_score', 'grade_marks_ratio', 'risk_index']
    missing_features = [f for f in engineered_features if f not in df.columns]
    if missing_features:
        print(f"\n⚠️ Missing engineered features: {missing_features}")
        print("Creating engineered features...")
        df['attendance_score'] = df['attendance'] / 100.0
        df['grade_marks_ratio'] = df['marks'] / np.maximum(df['grade'], 1)
        df['risk_index'] = ((100 - df['attendance']) * (100 - df['marks'])) / 100.0
        print("✅ Engineered features created")
    else:
        print("✅ All engineered features present")
    
    # Basic statistics
    print("\n📊 Basic Statistics:")
    print(df.describe())
    
except Exception as e:
    print(f"❌ Error loading dataset: {e}")
    raise
```

```python
# Prepare features for evaluation (UPDATED for new feature engineering)
print("🔧 Preparing features for evaluation...")

# Define feature groups
numerical_features = ['age', 'grade', 'attendance', 'marks']
engineered_features = ['attendance_score', 'grade_marks_ratio', 'risk_index']
target_column = 'future_performance'

# Get numerical features
numerical_data = df[numerical_features + engineered_features].values

# One-Hot encode subjects (IMPROVED from Label Encoding)
subjects = df['subject'].values.reshape(-1, 1)
subject_encoded = subject_encoder.transform(subjects)

# Combine features (same order as training)
X = np.hstack([numerical_data, subject_encoded])
y = df[target_column].values

print(f"✅ Features prepared: {X.shape[0]} samples, {X.shape[1]} features")
print(f"✅ Target prepared: {y.shape[0]} samples")
print(f"\n📋 Feature breakdown:")
print(f"  - Numerical features: {numerical_features}")
print(f"  - Engineered features: {engineered_features}")
print(f"  - One-Hot encoded subjects: {X.shape[1] - len(numerical_features) - len(engineered_features)} features")

# Create stratified bins for splitting (IMPROVED)
def create_performance_bins(y):
    bins = [0, 55, 75, 100]
    labels = ['low', 'average', 'high']
    return pd.cut(y, bins=bins, labels=labels, include_lowest=True)

y_bins = create_performance_bins(y)

# Split data with stratification (same parameters as training)
TEST_SIZE = 0.2
RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=y_bins
)

print(f"\n📊 Data Split (Stratified):")
print(f"Training set: {len(X_train)} samples ({(1-TEST_SIZE)*100:.0f}%)")
print(f"Test set: {len(X_test)} samples ({TEST_SIZE*100:.0f}%)")

# Scale features
X_test_scaled = scaler.transform(X_test)
X_train_scaled = scaler.transform(X_train)
print("✅ Features scaled for prediction")
```

## 3. Make Predictions

Use the trained model to make predictions on the test dataset.

```python
# Make predictions (with clamping to valid range)
print("🔮 Making predictions on test data...")

try:
    y_pred = model.predict(X_test_scaled)
    
    # Clamp predictions to valid range [0, 100] (SAFETY IMPROVEMENT)
    y_pred = np.clip(y_pred, 0, 100)
    
    print(f"✅ Predictions completed: {len(y_pred)} predictions generated")
    print("✅ Predictions clamped to [0, 100] range")
    
    # Basic prediction statistics
    print("\n📊 Prediction Statistics:")
    print(f"Actual values range: {y_test.min():.2f} - {y_test.max():.2f}")
    print(f"Predicted values range: {y_pred.min():.2f} - {y_pred.max():.2f}")
    print(f"Average actual: {y_test.mean():.2f}")
    print(f"Average predicted: {y_pred.mean():.2f}")
    
except Exception as e:
    print(f"❌ Error making predictions: {e}")
    raise
```

```python
# Create a DataFrame with actual vs predicted values
results_df = pd.DataFrame({
    'actual': y_test,
    'predicted': y_pred,
    'error': y_test - y_pred,
    'abs_error': np.abs(y_test - y_pred)
})

# Add performance categories
results_df['performance_category'] = pd.cut(
    results_df['actual'], 
    bins=[0, 40, 60, 80, 100], 
    labels=['Poor', 'Below Average', 'Average', 'Excellent']
)

print(" Results DataFrame created")
print(f"Shape: {results_df.shape}")
print("\n Sample Results:")
print(results_df.head(10))
```

## 4. Accuracy Metrics

Calculate comprehensive accuracy metrics to evaluate model performance.

```python
# Calculate comprehensive accuracy metrics
print("📊 Calculating accuracy metrics...")

# Regression metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
mape = mean_absolute_percentage_error(y_test, y_pred) * 100

# Adjusted R² (using feature_order which contains all feature names)
n = len(y_test)
p = len(feature_order)  # Fixed: use feature_order instead of feature_columns
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Custom accuracy metrics
threshold_5 = np.mean(np.abs(y_test - y_pred) <= 5.0) * 100
threshold_10 = np.mean(np.abs(y_test - y_pred) <= 10.0) * 100
threshold_15 = np.mean(np.abs(y_test - y_pred) <= 15.0) * 100

# Model evaluation based on R² threshold (target: 0.88)
target_r2 = 0.88
meets_target = r2 >= target_r2

print("\n" + "=" * 60)
print("📈 MODEL PERFORMANCE METRICS")
print("=" * 60)
print(f"\n📐 Regression Metrics:")
print(f"  • R² Score: {r2:.4f} {'✅ EXCEEDS TARGET' if meets_target else '❌ Below target'} (target: {target_r2})")
print(f"  • Adjusted R²: {adjusted_r2:.4f}")
print(f"  • RMSE: {rmse:.4f}")
print(f"  • MAE: {mae:.4f}")
print(f"  • MAPE: {mape:.2f}%")

print(f"\n🎯 Accuracy within Thresholds:")
print(f"  • Within ±5 points: {threshold_5:.2f}%")
print(f"  • Within ±10 points: {threshold_10:.2f}%")
print(f"  • Within ±15 points: {threshold_15:.2f}%")

print(f"\n🔧 Model Configuration:")
print(f"  • Algorithm: Random Forest Regressor")
print(f"  • Number of estimators: {model.n_estimators}")
print(f"  • Max depth: {model.max_depth}")
print(f"  • Number of features: {p}")
print("=" * 60)
```

```python
# Create metrics dictionary from computed values
metrics = {
    'r2_score': r2,
    'adjusted_r2': adjusted_r2,
    'mae': mae,
    'rmse': rmse,
    'mape': mape,
    'mse': mse,
    'accuracy_within_5': threshold_5,
    'accuracy_within_10': threshold_10,
    'accuracy_within_15': threshold_15,
    'total_predictions': len(y_pred)
}

# Display accuracy metrics in a formatted way
print("="*70)
print("🎯 MODEL ACCURACY METRICS (RandomForest - IMPROVED)")
print("="*70)

print("\n✅ ACCURACY METRICS:")
print("-"*70)
print(f"  R² Score (Coefficient of Determination):  {metrics['r2_score']:.4f} ({metrics['r2_score']*100:.2f}%)")
print(f"  Adjusted R² Score:                         {metrics['adjusted_r2']:.4f} ({metrics['adjusted_r2']*100:.2f}%)")
print(f"  Accuracy within ±5 points:                 {metrics['accuracy_within_5']:.2f}%")
print(f"  Accuracy within ±10 points:                {metrics['accuracy_within_10']:.2f}%")

print("\n📉 ERROR METRICS:")
print("-"*70)
print(f"  Mean Absolute Error (MAE):                 {metrics['mae']:.4f} points")
print(f"  Root Mean Squared Error (RMSE):            {metrics['rmse']:.4f} points")
print(f"  Mean Absolute Percentage Error (MAPE):     {metrics['mape']:.2f}%")

print("\n📊 PREDICTION STATISTICS:")
print("-"*70)
print(f"  Total test predictions:                    {metrics['total_predictions']}")
print(f"  Actual values range:                       {y_test.min():.2f} - {y_test.max():.2f}")
print(f"  Predicted values range:                    {y_pred.min():.2f} - {y_pred.max():.2f}")
print(f"  Average prediction error:                  ±{metrics['mae']:.2f} points")

# Performance interpretation
print("\n📋 INTERPRETATION:")
print("-"*70)

if metrics['r2_score'] >= 0.9:
    r2_quality = "🏆 Excellent"
elif metrics['r2_score'] >= 0.8:
    r2_quality = "✅ Very Good"
elif metrics['r2_score'] >= 0.7:
    r2_quality = "👍 Good"
elif metrics['r2_score'] >= 0.5:
    r2_quality = "⚠️ Moderate"
else:
    r2_quality = "❌ Needs Improvement"

print(f"  R² Score Quality: {r2_quality}")
print(f"  The model explains {metrics['r2_score']*100:.1f}% of the variance in student performance.")

if metrics['accuracy_within_5'] >= 80:
    print(f"  ✅ High accuracy: {metrics['accuracy_within_5']:.1f}% of predictions are within ±5 points")
elif metrics['accuracy_within_5'] >= 60:
    print(f"  👍 Moderate accuracy: {metrics['accuracy_within_5']:.1f}% of predictions are within ±5 points")
else:
    print(f"  ⚠️ Low accuracy: Only {metrics['accuracy_within_5']:.1f}% of predictions are within ±5 points")

print(f"  Average prediction is off by {metrics['mae']:.2f} points from actual performance.")

# Check if target achieved
print("\n🎯 TARGET ACHIEVEMENT:")
print("-"*70)
target_r2 = 0.88
if metrics['r2_score'] >= target_r2:
    print(f"  ✅ R² Target ACHIEVED: {metrics['r2_score']:.4f} ≥ {target_r2}")
else:
    print(f"  ❌ R² Target NOT MET: {metrics['r2_score']:.4f} < {target_r2}")
print("="*70)
```

```python
# Create an interactive prediction example (UPDATED for new model)
print("🔮 INTERACTIVE PREDICTION EXAMPLE")
print("="*60)

# Example student data
example_students = [
    {
        'name': 'High Performer',
        'age': 16,
        'grade': 11,
        'attendance': 95.0,
        'marks': 88.0,
        'subject': 'Mathematics'
    },
    {
        'name': 'Average Student',
        'age': 15,
        'grade': 10,
        'attendance': 75.0,
        'marks': 65.0,
        'subject': 'Science'
    },
    {
        'name': 'Struggling Student',
        'age': 17,
        'grade': 12,
        'attendance': 55.0,
        'marks': 45.0,
        'subject': 'English'
    },
    {
        'name': 'At-Risk Student',
        'age': 15,
        'grade': 10,
        'attendance': 40.0,
        'marks': 35.0,
        'subject': 'History'
    }
]

print("Predicting performance for example students (with 95% CI):")
print("-"*60)

def calculate_confidence_interval(prediction, attendance, marks):
    """Calculate 95% confidence interval for prediction"""
    base_std = 4.5
    attendance_factor = 1.0 + (1.0 - attendance / 100) * 0.5
    marks_deviation = abs(marks - 65) / 35
    marks_factor = 1.0 + marks_deviation * 0.3
    adjusted_std = base_std * attendance_factor * marks_factor
    margin = 1.96 * adjusted_std
    return max(0, prediction - margin), min(100, prediction + margin)

for student in example_students:
    # Create engineered features
    attendance_score = student['attendance'] / 100.0
    grade_marks_ratio = student['marks'] / max(student['grade'], 1)
    risk_index = ((100 - student['attendance']) * (100 - student['marks'])) / 100.0
    
    # Prepare numerical features
    numerical_data = [
        student['age'],
        student['grade'], 
        student['attendance'],
        student['marks'],
        attendance_score,
        grade_marks_ratio,
        risk_index
    ]
    
    # One-Hot encode subject
    subject_encoded = subject_encoder.transform([[student['subject']]])[0]
    
    # Combine features
    input_data = np.array([numerical_data + list(subject_encoded)])
    
    # Scale input
    input_scaled = scaler.transform(input_data)
    
    # Make prediction
    prediction = model.predict(input_scaled)[0]
    prediction = np.clip(prediction, 0, 100)  # Clamp to valid range
    
    # Calculate confidence interval
    lower, upper = calculate_confidence_interval(prediction, student['attendance'], student['marks'])
    
    print(f"\n👤 {student['name']}")
    print(f"   Current Marks: {student['marks']:.0f} | Attendance: {student['attendance']}%")
    print(f"   📊 Predicted Performance: {prediction:.1f}")
    print(f"   📈 95% Confidence Interval: [{lower:.1f}, {upper:.1f}]")
    print(f"   Risk Index: {risk_index:.2f}")
```
